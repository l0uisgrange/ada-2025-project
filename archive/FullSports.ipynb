### c'. Group subreddits using manual grouping

```python

# Define themes
THEMES = {
    'sports': [
        'nba', 'nfl', 'hockey', 'fantasyfootball', 'cfb', 'maddenultimateteam', 'baseball',
        'golf', 'patriots', 'collegebasketball', 'seahawks', 'minnesotavikings', 'madden',
        'greenbaypackers', 'warriors', 'eagles', 'denverbroncos', 'torontobluejays', '49ers',
        'lakers', 'browns', 'oaklandraiders'
    ]
}

# Extract sports theme list
sports_list = THEMES['sports']
# sports_list

# Combine all subreddit from dataset
all_subreddits = pd.concat([
    df_title["SOURCE_SUBREDDIT"],
    df_title["TARGET_SUBREDDIT"],
    df_body["SOURCE_SUBREDDIT"],
    df_body["TARGET_SUBREDDIT"]
]).unique()


# -------------------------------------------------- FIND SPORTS SUBREDDITS --------------------------------------------------

# Any keyword appears anywhere (nba --> nba, nba_fanpage --> nba, thenba --> nba)
# sports_present = [
#     s for s in all_subreddits
#     if any(keyword in s.lower() for keyword in sports_list)
# ]
# sports_present

# Whole-word match using regex (nba --> nba, nba_fanpage --> nba)
sports_present_cleaned = []
for s in all_subreddits:
    if not isinstance(s, str):
        continue
    name = s.lower()
    for keyword in sports_list:
        if re.search(rf"\b{keyword}\b", name):
            sports_present_cleaned.append(s)
            break
sports_present_cleaned

# Exact match (nba --> nba)
# sports_exact = [s for s in all_subreddits if s.lower() in sports_list]
# sports_exact


# -------------------------------------------------- SORTING & COUNTING --------------------------------------------------

# Sort + clean list
sorted(sports_present_cleaned)

# Count the activity of each sports subreddit
sports_counts = (
    pd.concat([
        df_title["SOURCE_SUBREDDIT"],
        df_title["TARGET_SUBREDDIT"],
        df_body["SOURCE_SUBREDDIT"],
        df_body["TARGET_SUBREDDIT"]
    ])
    .value_counts()
    .loc[sports_present_cleaned]
)
# sports_counts
```

### d. Plot external event dataset on global timeline

```python
import pandas as pd
import matplotlib.pyplot as plt

# Convert dates to datetime
df_english_pl['DateTime'] = pd.to_datetime(df_english_pl['DateTime'])
df_superbowl['Date'] = pd.to_datetime(df_superbowl['Date'])

# Define timeline
start_date = '2014-01-01'
end_date   = '2017-07-31'

# Filter events within timeline
df_matches_filtered = df_english_pl[(df_english_pl['DateTime'] >= start_date) & (df_english_pl['DateTime'] <= end_date)]
df_superbowl_filtered = df_superbowl[(df_superbowl['Date'] >= start_date) & (df_superbowl['Date'] <= end_date)]

# Number of matches per day
epl_count = df_matches_filtered.groupby(df_matches_filtered['DateTime'].dt.date).size().rename('EPL Matches')
sb_count = df_superbowl_filtered.groupby(df_superbowl_filtered['Date'].dt.date).size().rename('Super Bowl')

# Merge into one timeline
timeline = pd.concat([epl_count, sb_count], axis=1).fillna(0)
timeline.index = pd.to_datetime(timeline.index)  # convert index to datetime

plt.figure(figsize=(16,6))

plt.plot(timeline.index, timeline['EPL Matches'], label='EPL Matches per Day')
plt.plot(timeline.index, timeline['Super Bowl'], label='Super Bowl Event', marker='o', linestyle='None')

plt.xlabel("Date")
plt.ylabel("Number of Events")
plt.title("Timeline of EPL Matches and Super Bowl Events")
plt.legend()
plt.grid(True)
plt.show()
```

### c. Analyze interactions using LIWC + sentiment before/during/after events

```python
import networkx as nx

# -------------------------------------------------- SEGMENT DATASET BY TIME --------------------------------------------------

event_date = pd.to_datetime('2016-02-07')  # Super Bowl example
time_window = pd.Timedelta(days=7)

df_before = df_body[df_body['TIMESTAMP'] < event_date - time_window]
df_during = df_body[
    (df_body['TIMESTAMP'] >= event_date - time_window) &
    (df_body['TIMESTAMP'] <= event_date + time_window)
]
df_after = df_body[df_body['TIMESTAMP'] > event_date + time_window]


# -------------------------------------------------- AGGREGATE LIWC + SENTIMENT PER INTERACTION --------------------------------------------------

features = liwc_columns + ['VADER_POS', 'VADER_NEG', 'VADER_COMP']

agg_before = df_before.groupby(['SOURCE_SUBREDDIT', 'TARGET_SUBREDDIT'])[features].mean().reset_index()
agg_during = df_during.groupby(['SOURCE_SUBREDDIT', 'TARGET_SUBREDDIT'])[features].mean().reset_index()
agg_after = df_after.groupby(['SOURCE_SUBREDDIT', 'TARGET_SUBREDDIT'])[features].mean().reset_index()


# -------------------------------------------------- COMPARE INTERACTIONS ACROSS TIME --------------------------------------------------

# Mean positive emotion per segment
print("Mean LIWC_POSEMO")
print("Before:", agg_before['LIWC_POSEMO'].mean())
print("During:", agg_during['LIWC_POSEMO'].mean())
print("After:", agg_after['LIWC_POSEMO'].mean())

# Plot distribution of LIWC_POSEMO before/during/after
agg_before['segment'] = 'Before'
agg_during['segment'] = 'During'
agg_after['segment'] = 'After'
df_all = pd.concat([agg_before, agg_during, agg_after])
plt.figure(figsize=(8,5))
sns.boxplot(data=df_all, x='segment', y='LIWC_POSEMO')
plt.title('Distribution of Positive Emotion Across Event Segments')
plt.show()


# -------------------------------------------------- INTERACTION NETWORK WEIGHTED BY SENTIMENT --------------------------------------------------

G = nx.from_pandas_edgelist(
    agg_during,
    'SOURCE_SUBREDDIT',
    'TARGET_SUBREDDIT',
    edge_attr='LIWC_POSEMO',
    create_using=nx.DiGraph()
)

edge_colors = [G[u][v]['LIWC_POSEMO'] for u,v in G.edges()]
node_sizes = [G.degree(n)*200 for n in G.nodes()]

plt.figure(figsize=(12,10))
nx.draw(
    G,
    with_labels=True,
    node_size=node_sizes,
    edge_color=edge_colors,
    edge_cmap=plt.cm.Reds,
    node_color='skyblue'
)
plt.title('Interaction Network During Event (Edges = Positive Emotion)')
plt.show()
```
